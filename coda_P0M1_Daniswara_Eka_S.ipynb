```python
from selenium import webdriver
import requests
from bs4 import BeautifulSoup
```

```python
import pandas as pd
```

```python
import time
```

```python
driver = webdriver.Chrome()
# Deklarasi Variable
Nama_Produk = []
harga = []
image = []
link_produk = []


driver.get('https://www.kickavenue.com/search/sneakers?sex=M&sort_by=featured_item_score_desc&keyword=')
SCROLL_PAUSE_TIME = 1.2
# Get scroll height
last_height = driver.execute_script("return document.body.scrollHeight")

while True:
    # Scroll down to bottom
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

    # Wait to load page
    time.sleep(SCROLL_PAUSE_TIME)

    # Calculate new scroll height and compare with last scroll height
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height:
        break
    last_height = new_height
time.sleep(20)
html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')   



for data in soup.find_all('div', {'class':'sneaker col-6 col-sm-6 col-md-4 col-lg-3'}): 
    try:
        Nama_Produk.append(data.find('h2', {'class':'price-tag'}).get_text()) # mengambil semua data nama produk
    except:
        Nama_Produk.append(None) # diisi none jika tidak ada data
    try:
        harga.append(data.find('span', {'class':'currency-tag currency-tag__black text-left price-tag text-bold'}).get_text()) # mengambil semua data harga
    except:
        harga.append(None)  # diisi none jika tidak ada data
    try:
        image.append(data.find('img', {'class':'card-img'})['src']) # mengambil semua data image
    except:
        image.append(None)  # diisi none jika tidak ada data
    try:
        link_produk.append(data.find('a', {'target':'_self'})['href']) # mengambil semua data nama link produk
    except:
        link_produk.append(None)  # diisi none jika tidak ada data
        
```

```python
# membuat data frame
df_scrap = pd.DataFrame()

df_scrap['nama_produk'] = Nama_Produk # membuat kolom dari data Nama_Produk
df_scrap['harga'] = harga # membuat kolom dari data harga
df_scrap['gambar'] = image # membuat kolom dari data image
df_scrap['link_produk'] = link_produk # membuat kolom dari data link_produk
```

```python
# menghitung jumlah data pada masing2 kolom
print(len(Nama_Produk))
print(len(harga))
print(len(image))
print(len(link_produk))
```
